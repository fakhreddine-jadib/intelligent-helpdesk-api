```python
import sys
sys.path.append("..")
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split

PROC_DIR = Path("../data/processed")
GOLD_PATH = PROC_DIR / "gold_annotations.csv"

df = pd.read_csv(PROC_DIR / "tickets_clean_en.csv")

_, te_df = train_test_split(df, test_size=0.2, random_state=42,
                            stratify=df["queue"])
te_df = te_df.reset_index(drop=True)
print("Test set:", len(te_df))
```

```python
df = pd.read_csv(PROC_DIR / "tickets_clean_en.csv")
print(df.columns.tolist())
```

```python
print("te_df shape:", te_df.shape)
print("te_df columns:", list(te_df.columns))
print("\ndf columns:", list(df.columns))
```

```python
N_PER_CLASS = 30

parts = []
for q, group in te_df.groupby("queue"):
    parts.append(group.sample(min(len(group), N_PER_CLASS), random_state=7))

sample = pd.concat(parts).reset_index(drop=True)

print(sample["queue"].value_counts())
print("\nTotal to annotate:", len(sample))
print("Columns:", list(sample.columns))
```

```python
GOLD_PATH = PROC_DIR / "gold_annotations.csv"

gold = pd.DataFrame({
    "row_id": range(len(sample)),
    "text": sample["text"],
    "my_queue": "",
    "my_priority": "",
    "note": "",
})
gold.to_csv(GOLD_PATH, index=False, encoding="utf-8-sig")

# answer key — kept separate, do not open until annotation is finished
sample[["queue", "priority"]].to_csv(PROC_DIR / "gold_truth.csv", index=False)

print(f"Wrote {len(gold)} rows to {GOLD_PATH}")
print("Answer key saved separately to gold_truth.csv")
```
