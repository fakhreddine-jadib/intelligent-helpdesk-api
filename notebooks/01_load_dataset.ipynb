```python
import pandas as pd
from pathlib import Path
```

```python
RAW_DIR = Path("../data/raw")
print([f.name for f in RAW_DIR.iterdir() if f.is_file()])
```

```python
df4 = pd.read_csv(RAW_DIR / "dataset-tickets-multi-lang-4-20k.csv")
df3 = pd.read_csv(RAW_DIR / "dataset-tickets-multi-lang3-4k.csv")

for name, d in [("v4", df4), ("v3", df3)]:
    print(f"\n=== {name} === shape: {d.shape}")
    print("columns:", list(d.columns))
```

```python
# language availability — this decides the FR question
for name, d in [("v4", df4), ("v3", df3)]:
    print(f"\n=== {name} ===")
    print(d["language"].value_counts())
```

```python
# label distributions on the main corpus
print("QUEUE:\n", df4["queue"].value_counts(), "\n")
print("PRIORITY:\n", df4["priority"].value_counts(), "\n")
print("TYPE:\n", df4["type"].value_counts(), "\n")
if "business_type" in df4.columns:
    print("BUSINESS TYPE:\n", df4["business_type"].value_counts())
```

```python
# read three real tickets end to end
for _, row in df4.head(3).iterrows():
    print("\n" + "="*70)
    print(f"[{row['queue']} | priority {row['priority']} | {row['language']}]")
    print("SUBJECT:", row["subject"])
    print("BODY:", str(row["body"])[:600])
```
